## 📊 Numerical Feature Selection for Outlier Analysis — `store.csv`

The `store.csv` dataset contains several numerical columns related to store characteristics, competition, and promotional campaigns. However, **not every numerical column is suitable for continuous numerical outlier analysis**.

For the initial analysis, the focus will be on `CompetitionDistance`, as it represents a measurable numerical quantity.

### Feature Selection Criteria

| Feature                     | Data Type / Meaning                                    | Outlier Analysis                                |
| --------------------------- | ------------------------------------------------------ | ----------------------------------------------- |
| `Store`                     | Unique store identifier                                | Exclude — identifier, not a measurable quantity |
| `CompetitionDistance`       | Distance to the nearest competitor, measured in meters | **Include** — continuous numerical feature      |
| `CompetitionOpenSinceMonth` | Month when the nearest competitor opened               | Exclude — calendar component                    |
| `CompetitionOpenSinceYear`  | Year when the nearest competitor opened                | Exclude — calendar component                    |
| `Promo2SinceWeek`           | Calendar week when Promo2 started                      | Exclude — calendar component                    |
| `Promo2SinceYear`           | Year when Promo2 started                               | Exclude — calendar component                    |

### 🎯 Objective

The objective is to identify potential outliers in `CompetitionDistance` using statistical techniques such as the **Interquartile Range (IQR)** method.

The IQR method defines the lower and upper bounds as:

$$
\text{IQR}=Q_3-Q_1
$$

$$
\text{Lower Bound}=Q_1-1.5\times\text{IQR}
$$

$$
\text{Upper Bound}=Q_3+1.5\times\text{IQR}
$$

Observations outside these bounds will be flagged for further investigation.

### 💡 Key Consideration

An observation flagged as an outlier is not necessarily an error. A store may legitimately be located much farther from its nearest competitor than most other stores. Therefore, outliers should be investigated before deciding whether to retain, transform, or remove them.

**Conclusion:** `CompetitionDistance` will be the primary feature for the initial outlier analysis because it represents a continuous, measurable quantity. Other numerical columns will be excluded from this initial analysis because they represent identifiers or calendar components rather than continuous measurements.

````python
col = 'CompetitionDistance'

# Exclude missing values from the calculation
data = store_df[col].dropna()

# Calculate quartiles and IQR
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

# Calculate outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Count outliers
outlier_mask = (data < lower_bound) | (data > upper_bound)
outlier_count = outlier_mask.sum()

outlier_summary = pd.DataFrame({
    'Column': [col],
    'Q1': [Q1],
    'Q3': [Q3],
    'IQR': [IQR],
    'Lower_Bound': [lower_bound],
    'Upper_Bound': [upper_bound],
    'Outlier_Count': [outlier_count]
})

display(outlier_summary)
````

### 📊 Outlier Treatment — `CompetitionDistance`

The `CompetitionDistance` feature contains extreme values, with a maximum distance of **75,860 meters**. The Interquartile Range (IQR) method identified an upper boundary of **16,107.5 meters**.

**Methodology**

* Applied IQR-based capping to limit values exceeding the upper boundary.
* Retained all store records without deleting observations.
* Kept the lower boundary unchanged because distances are nonnegative and the calculated lower bound is negative.

| Statistical Measure | Before Capping |
| ------------------- | -------------: |
| Count               |          1,115 |
| Mean                |     5,396.61 m |
| Median              |        2,325 m |
| Maximum             |       75,860 m |

### 🏪 Missing Value Treatment — `Open`

The `Open` column indicates whether a store is operating (`1` = Open, `0` = Closed). The test dataset contains **11 missing values (0.03%)** in this feature.

**Methodology**

* Calculate the most frequent `Open` status for each **`Store` + `DayOfWeek`** combination using the training dataset.
* Use the corresponding store-and-weekday pattern to impute missing `Open` values in the test dataset.
* Investigate any combinations without sufficient historical information separately.

**💡 Justification:** Opening patterns may vary by store and day of the week. Using historical store-specific patterns is more representative than applying a single overall mode.

**✅ Conclusion:** Store-and-weekday-based imputation preserves historical operating patterns and provides a data-driven approach to handling the 11 missing `Open` values.


````python
print("Remaining missing Open values:",
      test_df['Open'].isnull().sum())
````

````python
outlier_cols = ['Sales', 'Customers']

outlier_results = []

for col in outlier_cols:
    values = train_df[col].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (values < lower_bound) |
        (values > upper_bound)
    ).sum()

    outlier_results.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower_Bound': lower_bound,
        'Upper_Bound': upper_bound,
        'Outlier_Count': outlier_count
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
````
